```python
#reusablefunction hain bar bar ka kam ek function mein
def clean_silver_layer(
    bronze_path,
    silver_path,
    quarantine_path,
    critical_col,
    fillna_col=None,
    fillna_value="NA"
):
    df = spark.read.format("parquet").load(bronze_path)

    df_dedup = df.dropDuplicates()

    df_good = df_dedup.filter(col(critical_col).isNotNull())
    df_quarantine = df_dedup.filter(col(critical_col).isNull())

    if fillna_col:
        df_final = df_good.fillna({fillna_col: fillna_value})
    else:
        df_final = df_good

    df_final.write.format("parquet").mode("overwrite").save(silver_path)

    df_quarantine.write.format("parquet").mode("overwrite").save(quarantine_path)

    return df_final, df_quarantine


# Function call — NO indentation
df_final, df_quarantine = clean_silver_layer(
    "/Volumes/workspace/default/bronze_practice/bronze/address",
    "/Volumes/workspace/default/bronze_practice/silver/address",
    "/Volumes/workspace/default/bronze_practice/quarantine/address",
    critical_col="add_id",
    fillna_col="ct_id"
)

# Output पाहण्यासाठी
df_final.show()
df_quarantine.show()

print("Silver count:", df_final.count())
print("Quarantine count:", df_quarantine.count())
```

```python
#explode hum silver mein karnge
from pyspark.sql.functions import explode, split

data = [(1, "Mary", "Prepaid,Postpaid"), (2, "John", "Prepaid")]
df_demo = spark.createDataFrame(data, ["sid", "name", "plans"])
df_demo.show()
```

```python
df_split = df_demo.withColumn("plans_array", split(df_demo["plans"], ","))
df_split.show(truncate=False)
```

```python
df_exploded = df_split.withColumn("plan_single", explode(df_split["plans_array"]))
df_exploded.show(truncate=False)
```

```python
from pyspark.sql.functions import explode_outer

data2 = [(1, "Mary", "Prepaid,Postpaid"), (2, "John", "Prepaid"), (3, "Susan", None)]
df_demo2 = spark.createDataFrame(data2, ["sid", "name", "plans"])

df_split2 = df_demo2.withColumn("plans_array", split(df_demo2["plans"], ","))

# Normal explode - Susan (NULL) gayab ho jayegi
print("Normal explode:")
df_split2.withColumn("plan_single", explode(df_split2["plans_array"])).show(truncate=False)

# explode_outer - Susan (NULL) rahegi, plan_single NULL dikhega
print("Explode outer:")
df_split2.withColumn("plan_single", explode_outer(df_split2["plans_array"])).show(truncate=False)
```
