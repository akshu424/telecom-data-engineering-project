```python
df_sb = spark.read.format("parquet").load("/Volumes/workspace/default/bronze_practice/gold/subscriber_details")

df_sb.createOrReplaceTempView("subscriber")

revenue_report = spark.sql("""
    SELECT 
        country AS Country,
        COUNT(subscriberid) AS total_subscriber,
        SUM(COALESCE(prepaid_amount, 0) + COALESCE(postpaid_amount, 0)) AS total_revenue
    FROM subscriber
    WHERE active_flag = 'A'
    GROUP BY country
""")

revenue_report.show()

revenue_report.write.format("csv").option("header", "true").mode("overwrite").save("/Volumes/workspace/default/bronze_practice/platinum/revenue_report")
```
